```python
import os
from dotenv import load_dotenv, find_dotenv
# find_dotenv() automatically locates .env in parent directories
load_dotenv(find_dotenv())
# Ensure both GEMINI_API_KEY and GOOGLE_API_KEY are available
os.environ["GEMINI_API_KEY"] = os.getenv("GEMINI_API_KEY")
```

```python
## Google Gemini
from langchain.chat_models import init_chat_model
model = init_chat_model("google_genai:gemini-flash-latest")
res = model.invoke("Hi how are you")
res.content
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-flash-latest")
res = model.invoke("What can you help me with")
print(res.content)
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:openai/gpt-oss-20b")
res = model.invoke("What can you help me with")
print(res.content)
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model = "openai/gpt-oss-120b")
res = model.invoke("What can you help me with")
print(res.content)
```

### Streaming and Batch

```python
model.invoke("Write me a 200 words paragraph on Artificial Inteliigence")
```

```python
## Streaming

for chunk in model.stream("Write me a 500 words paragraph on Aritificial Intelligence "):
    print(chunk.text , end = "|" , flush = True)
```

### Batch

```python
responses = model.batch([
    "why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "what is quantum computing?"
])

for response in responses :
    print(response)
```

```python
model.batch(
    [
        "why do parrots have colorful feathers?",
        "How do airplanes fly?",
        "what is quantum computing?",
    ],
    config={
        "max_concurrency": 5,
    }
)

```

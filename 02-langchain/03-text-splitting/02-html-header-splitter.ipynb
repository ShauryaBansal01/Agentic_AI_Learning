## How to split by HTML header
HTMLHeaderTextSplitter is a "structure-aware" chunker that splits text at the HTML element level and adds metadata for each header "relevant" to any given chunk. It can return chunks element by element or combine elements with the same metadata, with the objectives of (a) keeping related text grouped (more or less) semantically and (b) preserving context-rich information encoded in document structures. It can be used with other text splitters as part of a chunking pipeline.

```python
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string = """<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>My First Web Page</title>
</head>
<body>

    <!-- Main Heading -->
    <h1>Welcome to My Website</h1>

    <!-- Subheading -->
    <h2>About This Page</h2>

    <!-- Paragraph of text -->
    <p>This is a sample paragraph of text. HTML is used to structure the content of web pages.</p>

    <!-- Text formatting: Bold and Italic -->
    <p>You can make text <strong>bold</strong> or <em>italicized</em> easily.</p>

    <!-- Hyperlink to another website -->
    <p>Visit the official <a href="https://developer.mozilla.org" target="_blank">MDN Web Docs</a> to learn more.</p>

    <!-- Unordered List -->
    <h3>My Favorite Topics</h3>
    <ul>
        <li>Hypertext Markup Language (HTML)</li>
        <li>Cascading Style Sheets (CSS)</li>
        <li>JavaScript (JS)</li>
    </ul>

    <!-- Displaying an Image -->
    <h3>Sample Image</h3>
    <img src="https://placeholder.com" alt="A simple placeholder image" width="150" height="150">

</body>
</html>
"""


header_to_split = [("h1" , "Header1") , ("h2" , "Header2") , ("h3" , "Header3")]

html_splitter = HTMLHeaderTextSplitter(header_to_split)
html_header_split = html_splitter.split_text(html_string)
html_header_split
```

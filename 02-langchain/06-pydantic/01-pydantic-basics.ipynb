### Pydantic Basics: Creating and Using Models
Pydanti models are the foundation of data validation in Python. They use Python type annotations
to define the structure and validate data at runtime. Here's a detailed exploration of basic model
creation with several examples.

```python
from pydantic import BaseModel

class Person(BaseModel):
    name: str
    age: int
    city: str
    
person = Person(name="John Doe", age=30, city="New York")
print(person)
print(type(person))
```

```python
person1 = Person(name="John Doe", age=30, city=12)
print(person1)
```

```python
## Model With Optional Fields
from typing import Optional

class Employee(BaseModel):
    id: int 
    name:str
    department: str
    salary: Optional[float] = None  # Optional field with default value None
    is_active: Optional[bool] = True  # Optional field with default value True
```

```python
emp1 = Employee(id=1, name="Alice", department="HR")
print(emp1)
```

```python
class skill(BaseModel):
    skills:list[str]

skills = skill(skills=["Python", "Java", "C++"])
print(skills)
```

## Nested Model

```python
from pydantic import BaseModel


class Address(BaseModel):
    street: str
    city: str
    zip_code: str


class Customer(BaseModel):
    customer_id: int
    name: str
    address: Address


customer = Customer(
    customer_id=1,
    name="Emma",
    address={
        "street": "123 Main St",
        "city": "Boston",
        "zip_code": "02108",
    },
)

print(customer)
print(customer.address)
```

## Pydantic Field : Customization and Constraints

The Field function in Pydantic enhances model  fields beyond basic ftype hints by allowin
ou to specify validation rules default values, aliases, andi more. Here's a comprehensive tutorial with examples.

```python
from pydantic import BaseModel, Field
class Item(BaseModel):
    name: str = Field(min_length=1, max_length=100)
    price: float = Field(gt = 0 , le = 1000)
    quantity: int = Field(gt = 0)
```

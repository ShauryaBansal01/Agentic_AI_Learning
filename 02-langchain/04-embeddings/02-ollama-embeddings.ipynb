```python
from langchain_ollama import OllamaEmbeddings
```

```python
# gemma:2b is a chat model — it has no embedding head, so Ollama returns
# HTTP 500 "This server does not support embeddings".
# Use a dedicated embedding model instead (pull it once: `ollama pull nomic-embed-text`).
embeddings = OllamaEmbeddings(
    model="nomic-embed-text"
)
```

```python
r1 = embeddings.embed_documents(["Alpha is the first letter of the Greek alphabet. It is often used to denote the first or most significant item in a series."])
```

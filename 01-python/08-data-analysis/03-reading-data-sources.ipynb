## Reading Data From Different Sources

```python
import pandas as pd 
from io import StringIO
Data = '{"employee_name": "James" , "email": "james@example.com" , "job_profile": [{"title": "Team Lead" , "title2": "Sr developer"}]}'
df = pd.read_json(StringIO(Data))
```

```python
df
```

```python
df.to_json()
```

```python
df.to_json(orient='index')
```

```python
df.to_json(orient='records')
```

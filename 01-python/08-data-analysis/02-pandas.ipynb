### Pandas-DataFrame And Series
Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It
provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object,
while a DataFrame is a two-dimensional, size-mutable, and potewtially heterogeneous tabular data structure
with labeled axes (rows and columns).

```python
import pandas as pd

## Series - is a one dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). The axis labels are collectively referred to as the index. A Series is like a fixed-size dict in that you can get and set values by index label.

data = [1, 2, 3, 4, 5]
series = pd.Series(data)
print(series)
print(type(series))
```

```python
## Create Series from a dictionary 

data_dict = {'a': 1, 'b': 2, 'c': 3}
series_dict = pd.Series(data_dict)
print(series_dict)
print(type(series_dict))
```

```python
data = [10 , 20, 30, 40, 50]
## Alternative way - index = [1 , 2 , 3 , 4 , 5]
## series = pd.Series(data, index=index)
series = pd.Series(data, index=[1, 2, 4, 5, 6])
print(series)
```

```python
## DataFrame 
## Create a dataframe from a dictionary of lists

data = {
    "Name": ["Shaurya" , "Rohit" , "Ramesh" , "Suresh"],
    "Age": [25, 30, 35, 40],
    "City": ["New York", "Los Angeles", "Chicago", "Houston"]
}

df = pd.DataFrame(data , index = [1 , 2 , 3 ,4])
print(df)
```

```python
## Create a datafram from a list of dictionaries

data = [
    {"Name": "Shaurya", "Age": 25, "City": "New York"},
    {"Name": "Rohit", "Age": 30, "City": "Los Angeles"},
    {"Name": "Ramesh", "Age": 35, "City": "Chicago"},
    {"Name": "Suresh", "Age": 40, "City": "Houston"}
]

df = pd.DataFrame(data , index = [1 , 2 , 3 ,4])
print(df)
```

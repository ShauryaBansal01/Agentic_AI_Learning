### Data Manipulation and Analysis with Pandas

Data manipulation and analysis are key tasks in any data science or data analysis project. Pandas provides a
wide range of functions for data manipulation and analysis, making it easier to clean, transform, and extract
insights from data. In this lesson, we will cover various data manipulation and analysis techniques using Pandas.

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/data.csv')
## fetch the first 5 rows of the dataframe
df.head(5)
```

```python
df.tail(5) ## fetch the last 5 rows of the dataframe
```

```python
df.describe() ## Summary statistics of the dataframe
```

```python
df.dtypes
```

```python
## Handling missing values

df.isnull().any() ## Check if any row has missing values
```

```python
df.isnull().sum() ## Count the number of missing values in each column
```

```python
df.fillna(0) ## Fill missing values with 0
```

```python
df['rev'] = df['Revenue'].fillna(df['Revenue'].mean()) ## Fill missing values in 'rev' column with mean of the column
```

```python
df
```

```python
df.dtypes
```

```python
## Renaming column 
df.rename(columns={'rev':'main'}, inplace=True) ## Rename the column 'Revenue' to 'rev'
```

```python
df
```

```python
## Change Data type of a column

df['main'] = df['main'].astype(int) ## Change the data type of the column 'main' to int
df
```

```python
df['new'] = df.iloc[:, 8].apply(lambda x: x * 2)
df.head(5)
```

```python
## Data Aggregating and grouping 
df.head(5)
```

```python
grouped_mean = df.groupby('Product')['main'].mean() ## Group by 'Product' and calculate the mean of 'main' column
grouped_mean
```

```python
grouped_agg = df.groupby('Category')['main'].agg(['mean','sum','count'])
grouped_agg
```

```python
### Merging and joining dataframes

df1 = pd.DataFrame({'key': ['A', 'B', 'C', 'D' , 'E'], 'value1': [1, 2, 3, 4 , 5]})
df2 = pd.DataFrame({'key': ['A', 'B', 'C', 'D'], 'value2': [5, 6, 7, 8]})

```

```python
df1
```

```python
df2
```

```python
### Merge Dataframe on the key column

pd.merge(df1, df2, on='key',how="inner") ## Merge the two dataframes on the 'key' column
```

```python

pd.merge(df1, df2, on='key',how="outer") 
```

```python

pd.merge(df1, df2, on='key',how="left") 
```

```python

pd.merge(df1, df2, on='key',how="right") 
```

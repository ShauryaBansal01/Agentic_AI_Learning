### Magic Methods
Magic methods in Python, also known as dunder methods (double underscore methods), are special method
that start and end with double underscores. These methods enable you to define the behavior of objects for
built-in operations, such as arithmetic operations, comparisons, and more.
#### Magic Methods
Magic methods are predefined methods in Python that you can override to change the behavior of your
objects. Some common magic methods include:



```python
'''
__init__ - initialize the new instance of the class
__str__ - return a string representation of the object
__repr__ - return a string representation of the object for debugging
__len__ - return the length of the object
__getitem__ - retrieve an item from the object using indexing
__setitem__ - set an item in the object using indexing

'''
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Person(name={self.name}, age={self.age})"

    def __repr__(self):
        return f"Person(name={self.name!r}, age={self.age!r})"

    def __len__(self):
        return len(self.name)

    def __getitem__(self, index):
        return self.name[index]

    def __setitem__(self, index, value):
        name_list = list(self.name)
        name_list[index] = value
        self.name = ''.join(name_list)

person = Person("Alice", 30)
print(dir(Person))
print(person[0])  # Output: A
print(len(person))  # Output: 5
print(person)
```

### Decorators

Decorators are a powerful and flexible feature in Python that allows you to modify the behavior of a function or
class method. They are commonly used to add functionality to functions or methods without modifying thei
actual code. This lesson covers the basics of decorators, including how to create and use them.

```python
## Function Copy

def Welcome():
    return "Welcome to Python Programming"

Welcome()
```

```python
wel = Welcome
wel()
 
```

```python
## cLosures - function inside function  
def main_welcome(msg):
    def sub_welcome():
        print(msg)
        print("Welcome to inner function")
        print("Please learn these concepts properly")
    return sub_welcome()

main_welcome("Hello Python Learners")
```

```python
## Decorators - function as an argument to another function
def main_funtion(func):
    def sub_function():
        print("This is a sub function")
        func()
    return sub_function()
```

```python
def course_introduction():
    print("Please learn these concepts properly")

main_funtion(course_introduction)
```

```python
## Decorators - function as an argument to another function
@main_funtion
def course_introduction():
    print("Please learn these concepts properly")
```

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def greet(name):
    print(f"Hello, {name}!")

```

```python
greet("Alice")
```

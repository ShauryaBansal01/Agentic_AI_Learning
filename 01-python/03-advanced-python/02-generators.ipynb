### Generators
Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily,
which means they generate values on the fly and do not store them in memory.

````python
def square(n):
    for i in range(3):
        yield i ** 2


for i in square(3):
    print(i)
````

### Generator Notes

- Generators are functions that use `yield` instead of `return`.
- They produce values one at a time, only when needed.
- This makes them memory efficient for large sequences.
- Each `yield` pauses the function and remembers its state.
- When the next value is requested, execution continues from the last `yield`.

#### Why use generators?

- Less memory usage
- Faster startup for large data streams
- Useful for reading files, infinite sequences, or lazy calculations

#### Basic syntax

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

#### Example

```python
def square(n):
    for i in range(n):
        yield i * i

for value in square(5):
    print(value)
```

#### Common points to remember

- A generator does not return all values at once.
- It can be used in a `for` loop just like a list or iterator.
- After it is exhausted, it cannot be reused unless created again.
- `yield` and `return` are different: `yield` gives one value and pauses, while `return` ends the function.

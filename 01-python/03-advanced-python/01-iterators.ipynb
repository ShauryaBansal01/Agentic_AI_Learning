### Iterators
Iterators are advanced Python concepts that allow for efficient looping and memory management. Iterators
provide a way to access elements of collection sequentially without exposing the underlying structure.

```python
my_list = [1, 2, 3, 4, 5, 6]
for i in my_list:
    print(i)
```

```python
## Lazy Loading is a pattern that defers the initialization or loading of a resource until the exact moment it is needed. Used in the iterator 
iterator = iter(my_list)
```

```python

try:
    print(next(iterator))
except StopIteration:
    print("No more elements in the iterator.")
```

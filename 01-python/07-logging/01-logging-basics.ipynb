## Python Logging
Logging is a crucial aspect of any application, providing a way to track events, errors, and operational
information. Pythonis built-in logging module offers a flexible framework for emitting log messages from
Python programs. In this lesson, we will cover the basics of logging, including how to configure logging, log
levels, and best practices for using logging in Python applications.

```python
import logging

## Configuring the logging
logging.basicConfig(level = logging.DEBUG)

## log messages with different severity levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

### Log Levels

Python's `logging` module has several log levels that indicate the severity of events. The default levels are:

- **DEBUG**
  - Detailed information.
  - Typically useful only when diagnosing or debugging problems.

- **INFO**
  - Confirms that things are working as expected.
  - Used to record normal program execution.

- **WARNING**
  - Indicates that something unexpected happened.
  - Warns about a potential problem in the near future (e.g., *"disk space low"*).
  - The program continues to work as expected.

- **ERROR**
  - Indicates a more serious problem.
  - The software could not perform a specific function.

- **CRITICAL**
  - Represents a very serious error.
  - Indicates that the program itself may be unable to continue running.

```python
    ## configuring the logging 
import logging
logging.basicConfig(
    filename = 'app.log',
    filemode = 'w',
    level = logging.DEBUG,
    format = '%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt = "%Y-%m-%d %H:%M:%S",
)

logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

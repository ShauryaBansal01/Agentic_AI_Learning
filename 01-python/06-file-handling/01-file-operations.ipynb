### File Operation- Read And Write Files
File handling is a crucial part of any programming language. Python provides built-in functions and methods to
read from and write to files, both text and binary. This lesson will cover the basics of file handling, including
reading and writing text files and binary files.

```python
## Read a whole file All with respect to text file for now 
with open('../data/test.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
## read a file line by line 
with open('../data/test.txt' , 'r') as file:
    for line in file:
        print(line.strip())  # strip() removes the newline character
```

```python
with open('../data/test.txt' , 'w+') as file: ## will overwrite the file 
    file.write("Some text to write to the file.")
    file.write("\nAnother line of text.")
```

```python
with open('../data/test.txt' , 'a') as file: ## Append to a file
    file.write("\nSome text to write to the file.")
    file.write("\nAnother line of text.")
```

```python
lines = ["\nFirst line\n", "Second line\n", "Third line\n"]
with open('../data/test.txt', 'a') as file:
    file.writelines(lines)  # Write multiple lines to the file
```

```python
### Binary Files now 

data = b'\x00\x01\x02\x03\x04\x05'  # Binary data
with open('../data/test.bin', 'wb') as file:
    file.write(data)  # Write binary data to the file
```

```python
with open('../data/test.bin', 'rb') as file:
    binary_content = file.read()
    print(binary_content)  # Output: b'\x00\x01\x02\x03\x04\x05'
```

```python
# Read a text file and count the number of lines, words, and characters
def count_text_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = '../data/test.txt'
lines, words, characters = count_text_file(file_path)
print(f'Lines: {lines}, Words: {words}, Characters: {characters}')
```

```python
## writing and reading a text file 
with open('../data/test.txt', 'w+') as file:
    file.write("This is a sample text file.\n")
    file.write("It contains multiple lines of text.\n")
    
    file.seek(0)  # Move the cursor to the beginning of the file
    content = file.read()
    print(content)
    
```

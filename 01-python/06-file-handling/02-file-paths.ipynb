```python
import os 
cwd = os.getcwd()
print(cwd)
```

```python
## Create a new directory

new_directory = "Testdir"
os.mkdir(new_directory)
print(f"Directory '{new_directory}' created.")
```

```python
## List all files and directories in the current working directory
items = os.listdir(cwd)
print(items)
```

```python
## Join paths to create a full path to a file in the new directory
dir_name = "Testdir"
file_name = "testfile.txt"
full_path = os.path.join(dir_name, file_name)
print(f"Full path to the file: {full_path}")
```

```python
## Join paths to create a full path to a file in the new directory - in current working directory
dir_name = "Testdir"
file_name = "testfile.txt"
full_path = os.path.join(cwd ,dir_name, file_name)
print(f"Full path to the file: {full_path}")
```

```python
path = '../data/test.txt'
if os.path.exists(path):
    print(f"The file '{path}' exists.")
else :
    print(f"The file '{path}' does not exist.")
```

```python
path = '../data/example.csv'
if os.path.isfile(path):
    print(f"The path '{path}' is a file.")
elif os.path.isdir(path):
    print(f"The path '{path}' is a directory.")
else :
    print(f"The path '{path}' does not exist.")
```

```python
##getting the absolute path of a file
path = '../data/test.txt'
abosolute_path = os.path.abspath(path)
print(f"The absolute path of '{path}' is: {abosolute_path}")
 
```

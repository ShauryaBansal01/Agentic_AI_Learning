## DataTypes
### 1. Definition:

• Data types are a class'fication of data which tell the compiler or interpreter how the programmer intends
to use the data.

• They determine the type of operations that can be performed on the data, the values that the data can
take, and the amount of memory needed to store the data.


### 2. Importance of Data Types in Programming

Explanation:
• Data types ensure that data is stored in an efficient way.
e They help in performing correct operations on data.
• Proper use of data types can prevent errors and bugs in the program.

```python
name = "Shaurya"
age = 21 
cgpa = 8.53
single = True

print(name)
print(age)
print(cgpa)
print(single)


print(type(name))
print(type(age))
print(type(cgpa))
print(type(single))
```

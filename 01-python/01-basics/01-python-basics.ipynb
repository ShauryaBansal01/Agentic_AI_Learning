Indentation in Python is used to define the structure and hierarchy of the code. Unlike many other programming languages that use
braces {} to delimit blocks of code, Python uses indentation to determine the grouping of statements. This means that all the statemen
within a block must be indented at the same level.

```python
#Indentation 

age = 31

if age >= 30:
    print("Age is older than 30")
else :
    print("I cant print anything")
```

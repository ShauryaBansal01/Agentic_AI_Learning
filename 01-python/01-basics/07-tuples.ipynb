```python
#A tuple is an ordered collection of elements. Unlike lists, a tuple is immutable, meaning you cannot add, remove, or modify its elements.

tup = (1, 2, 3, 4, 5)
print(tup[0]) #accessing elements using index
print(tup[-1]) #accessing elements using negative index
```

```python
#single element tuple is trated as an integer by the python
t = (5)
print(type(t))
```

```python
t = (1,2,2,3,2)

print(t.count(2))
```

```python
t = (10,20,30)

print(t.index(20))
```

```python
#tuple packing 

student = "Rahul",21,"Delhi"

print(student)

# and unpacking
student = ("Rahul",21,"Delhi")

name, age, city = student

print(name)
print(age)
print(city)
```

```python
#Swapping Variables
a = 10
b = 20

a, b = b, a

print(a,b)
```

```python
#nested tuple

student = (
    "Rahul",
    (90,85,88)
)

print(student[1][0])
```

```python
#TUPLE CONCATENATION

t1 = (1,2)
t2 = (3,4)

t3 = t1 + t2

print(t3)
```

```python
#tuple repetition
t = (1,2)

print(t*3)
```

```python
t = (10,20,30,11)
print(20 in t)
#sorting a tuple returns a list
print(sorted(t))
```

```python
#tuple -> list conversion
t = (1,2,3)

lst = list(t)

#list -> tuple conversion
lst = [1,2,3]

t = tuple(lst)
```

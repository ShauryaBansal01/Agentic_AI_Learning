```python
lst = [1,2,3,4,5,6,7]
n = len(lst)

for i in range(0,n):
    print(lst[i])
```

```python
lst = []
for i in range(0,10):
    n = int(input("Enter a number"))
    lst.append(n)

for i in range(len(lst)):
    print(lst[i])
    if lst[i] == 5:
        lst.remove(lst[i])

for i in range(len(lst)):
    print(lst[i])
```

```python
# List comprehension

evensq = [i*i for i in range(1,11) if i%2==0]
print(evensq)
```

```python
lst = [1,2,3,4,5,6,7]
for i, value in enumerate(lst):
    print(f"Index: {i}, Value: {value}")    
```

```python
lst = []

for i in range(3):
    lst.append([])          # Create an empty row
    for j in range(3):
        n = int(input("Enter a number: "))
        lst[i].append(n)
        
for i in range(0,3):
    for j in range(0,3):
        print(lst[i][j])
```

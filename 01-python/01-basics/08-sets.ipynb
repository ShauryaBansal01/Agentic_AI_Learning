```python
#A set is an unordered collection of unique elements
number = {1, 2, 3, 4, 5}
print(number)
```

```python
numbers = {1,2,3,3,4,4,5,5}
print(numbers)
```

```python
names = ["Rahul","Aman","Rahul","Priya","Aman"]

unique_names = set(names)

print(unique_names)
```

```python
s = {10,20,30}

for x in s:
    print(x)
```

```python
s = {1,2,3}

s.add(4)

print(s)
```

```python
s = {1,2,3}

s.update([4,5,6])

print(s)

```

```python
s = {1,2,3}

s.remove(2)

print(s)
```

```python
s = {10,20,30}

x = s.pop()

print(x)
print(s)
```

```python
s = {10,20,30}

print(20 in s)
```

```python
A = {1,2,3,4}
B = {3,4,5,6}
print(A | B)
print(A.union(B))
print(A & B)
print(A.intersection(B))
print(A - B)
print(A ^ B)
```

```python
A = {1,2}
B = {1,2,3,4}
print(A.issubset(B))
B.issuperset(A)
```

```python
A = {1,2}
B = {3,4}

print(A.isdisjoint(B))
```

```python
#REMOVE DUPLICATES FROM A LIST

nums = [1,2,2,3,4,4]

unique = list(set(nums))
print(unique)
```

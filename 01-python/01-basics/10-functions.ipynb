```python
#A function is simply a block of reusable code that performs a specific task.

def area(length , breadth):
    return length * breadth


print(area(10,20))
```

```python
# Default Parameters

def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Shaurya")
```

```python
#Keyword Arguments

def student(name, age):
    print(name)
    print(age)

student(age=20, name="Shaurya")


#The order doesn't matter when using keyword arguments.
```

```python
#Accepts multiple positional arguments as a tuple.
def total(*numbers):
    total = 0
    for num in numbers:
        total += num
    print(total)

total(1, 2, 3, 4, 8)
```

```python
#kwargs
#Accepts multiple keyword arguments as a dictionary.

def student(**info):
    print(info)

student(name="Shaurya", age=20)
```

```python
# A lambda function is simply an anonymous (unnamed) function
# lambda arguments : expression

square = lambda x: x ** 2
print(square(5))  # Output: 25

multiply = lambda a, b: a*b
print(multiply(5,6))
```

Why Lambda?

Suppose you only need a function once.

Instead of writing

def cube(x):
    return x**3

you can write

lambda x: x**3

```python
#map function - map() applies a function to every element of an iterable.

numbers = [1,2,3,4]

result = map(lambda x: x*x, numbers)

print(list(result))
```

```python
def square(x):
    return x*x

numbers = [1,2,3,4]

result = map(square, numbers)

print(list(result))
```

```python
names = ["rahul","aman","priya"]

upper = map(str.upper, names)

print(list(upper))
```

```python
#Filter function - filter() filters elements from an iterable based on a function that returns True or False.

numbers = [1,2,3,4,5,6]
result = filter(lambda x: x%2==0, numbers)
print(list(result))
```

```python
numbers = [5,12,18,3,20] 
result = filter(lambda x: x > 10 , numbers)
print(list(result))
```

```python
#combining map and filter

numbers = [1,2,3,4,5,6]

even = filter(lambda x: x%2==0, numbers)

square = map(lambda x: x*x, even)

print(list(square))
```

```python
#Dictionary In Python - A dictionary stores data as key-value pairs.

student = {
    "name": "Shaurya",
    "age": 21 ,
    "isPlaced": False 
}

print(student)

```

```python
student = dict(name="Shaurya", age=21, isPlaced=False)
print(student)
print(student["name"]) #accessing value using key
print(student.get("cgpa"))
#empty dictionary
students = {}
students["name"] = "mishti"
print(students)
```

```python
person = {
    "name": "Shaurya",
    "address": {
        "street": "123 Main St",
        "city": "New York",
    }
}

print(person["address"]["city"])
```

```python
student = {
    "name":"Rahul",
    "age":21,
    "marks": 90
}

for key in student:
    print(key)
    
for value in student.values():
    print(value)


print("name" in student)
student.pop("age")
student.items()
student.popitem()
```

```python
for key, value in student.items():
    print(key, value)
    
student.keys()
student.values()
```

```python
#Dictionary Comprehension - A dictionary comprehension is a concise way to create dictionaries.
square = {x:x*x for x in range(5) if x%2==0}

print(square)
```

```python
d1 = {"a":1}
d2 = {"b":2}

d3 = d1 | d2

print(d3)
```

```python
arr = [1,2,2,3,3,3]

freq = {}

for x in arr:
    if x in freq:
        freq[x] += 1
    else:
        freq[x] = 1

print(freq)

freqq = {}
#shorter way
for x in arr:
    freqq[x] = freqq.get(x, 0) + 1
    
print(freqq)
```

```python
d = {
    "a":1,
    "b":2
}

reverse = {
    v:k
    for k,v in d.items()
}

print(reverse)
```

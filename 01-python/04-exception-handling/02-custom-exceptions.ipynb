```python
class Error(Exception):
    pass

class dobExpection(Error):
    pass

years = int(input("Enter your DOB "))
age = 2024 - years

try:
    if 20 >= age <= 30:
        print("You are eligible for this program")
    else:
        raise dobExpection()
except dobExpection:
    print("See the error message above.")
```
